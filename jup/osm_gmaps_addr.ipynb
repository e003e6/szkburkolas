```python
def osm_addr(lat, lon):
    url = f"https://nominatim.openstreetmap.org/reverse?format=json&lat={lat}&lon={lon}&addressdetails=1"

    headers = {"User-Agent": "MyGeocoderApp/1.0 (hello@hello.com)"}

    response = requests.get(url, headers=headers)
    data = response.json()
    
    return data
```

```python
def gmaps_address(lat, lon):
    '''
    lat: 47.
    lon: 18.
    '''
    url = f'https://www.google.hu/maps/preview/reveal?authuser=0&hl=hu&gl=hu&pb=!2m12!1m3!1d339.8894952009682!2d{lon}!3d{lat}!2m3!1f0!2f0!3f0!3m2!1i1512!2i386!4f13.1!3m2!2d{lon}!3d{lat}!4m2!1sdHq3Z83_IYim1fIPlKOvyAQ!7e81!5m5!2m4!1i96!2i64!3i1!4i8'
    response = requests.get(url)

    if response.status_code == 200:
        text = response.text
        data = text.strip()[4:]
        data = json.loads(data)

        cim = data[0][0].split()
        cim = (' '.join(cim[:-2]), cim[-2], cim[-1])
        
        adatok = []
        adatok.extend(cim)
        adatok.extend(data[0][1].split())
        adatok.extend(data[1][2][2:])

        return adatok
        
    else:
        print(f"Hiba: {response.status_code}")
```

```python
lat = 47.337563209823344
lon = 19.88764148270365
```

```python
gmaps_address(lat, lon)
```
